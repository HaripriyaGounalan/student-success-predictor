```python
# Observe the dataset
skim(structured_df)
```

```python
# Unique values in each column with value in  detail
for col in structured_df.columns:
    unique_vals = structured_df[col].dropna().unique()
    print(f"{col}: {len(unique_vals)} unique values")
    for val in unique_vals:
        count = structured_df[col].eq(val).sum()
        print(f"  {val}: {count}")
```
